# German Credit: DiCE Baseline Audit

This notebook uses the official `dice-ml` implementation to generate one counterfactual for each sample predicted as unfavorable; samples already assigned to the favorable class remain unchanged. The generated samples are recombined into an empirical distribution of 100 individuals and evaluated using coverage, SWD, WD, and four institutional constraint metrics.

```python
import sys
from pathlib import Path
import pandas as pd
from IPython.display import display

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / "explainers").is_dir() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate repository root.")

ROOT = find_repo_root()
BASELINE_DIR = ROOT / "baselines" / "dice"
OUTPUT_DIR = BASELINE_DIR / "results" / "german-credit"
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(BASELINE_DIR))
from dice_audit import run_german
print(f"Repository: {ROOT}")
```

## Run DiCE

DiCE does not receive Mean, Std., LSC, or FSD constraints. These metrics are audited only after the instance-level counterfactuals have been aggregated, allowing us to test whether a classical individual-level CE method naturally satisfies institution-level distributional requirements.

```python
runs, summary = run_german(ROOT, OUTPUT_DIR)
display(runs.round(5))
display(summary.round(5))
```
