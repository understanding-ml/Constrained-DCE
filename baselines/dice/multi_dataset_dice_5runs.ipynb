# Cardio, HELOC, and Market: Five-Run DiCE Audit

This notebook runs the official DiCE implementation on the three datasets using seeds `[40, 41, 42, 43, 44]`. One counterfactual is generated independently for each unfavorable sample; the results are then aggregated and evaluated for distributional distance, four institutional constraints, and data-domain validity.

```python
import sys
from pathlib import Path
import pandas as pd
from IPython.display import display

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / "explainers").is_dir() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate repository root.")

ROOT = find_repo_root()
BASELINE_DIR = ROOT / "baselines" / "dice"
OUTPUT_DIR = BASELINE_DIR / "results" / "multi-dataset"
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(BASELINE_DIR))
from dice_audit import SEEDS, run_other_datasets
print(f"Seeds: {SEEDS}")
```

## Five Runs

DiCE uses the same train/test split, MLP, and 100 explained samples. The five seeds control only counterfactual candidate sampling; they do not change the trained model or explained population. Mean is computed on the original feature scale. Consistent with the original DCE/CDCE validation notebooks, Std. uses the sample standard deviation (`ddof=1`). LSC uses the dataset-specific linear relation estimated from the training data, and FSD uses the same empirical-CDF violation-count definition.

```python
runs, summary = run_other_datasets(ROOT, OUTPUT_DIR)
display(runs.round(5))
display(summary.round(5))
```
